## 1. What Are Artificial Intelligence and Machine Learning?

### 1.1 Artificial intelligence: input → AI → output

Artificial intelligence (AI) is the broad goal of building systems that produce useful outputs—such as predictions, decisions, recommendations or actions—from inputs. The mechanism inside the system does not have to learn from data: it may be a formula, a collection of programmed rules, an expert system, or a machine-learning model.

![Input data passes through an AI system built from formulas and rules, an expert system, or machine learning to produce an output](assets/ai-input-output.svg){fig-alt="Input data passes through an AI system built from formulas and rules, an expert system, or machine learning to produce a prediction, decision, recommendation or action" style="width: 100%; height: auto;"}

### 1.2 Machine learning: learning the relationship from history

Machine learning (ML) is one way to build the AI system. Instead of manually writing every input-output rule, practitioners design or select a **method/algorithm** that learns a useful relationship from historical examples. Training converts those examples into **model parameters**; inference uses the learnt parameters to produce an output for a new input.

![Historical input-output examples pass through a learning algorithm to create model parameters, which a trained model uses to map a new input to an output](assets/ml-learning-from-history.svg){fig-alt="Historical input-output examples pass through a learning algorithm to create model parameters, which a trained model uses to map a new input to an output" style="width: 100%; height: auto;"}


## 2. From Machine Learning to Deep Learning

### 2.1 Why add more layers?

A single layer can capture a limited transformation. Multiple layers can build a hierarchy: early layers detect simple patterns, middle layers combine them, and later layers form task-specific concepts. The CNN below illustrates the progression from edges and textures to outlines, shapes and a final class prediction.

![A dog image passes through CNN layers that learn edges, outlines, textures and shapes before a classifier predicts dog](assets/cnn-dog-feature-extraction.png){fig-alt="A dog image passes through CNN layers that learn edges, outlines, textures and shapes before a classifier predicts dog" style="width: 100%; height: auto;"}

- **Early layers:** local edges and colour changes
- **Middle layers:** textures, contours and reusable parts
- **Later layers:** whole-object evidence and a task output

### 2.2 Learning from history as lossy compression

One useful viewpoint is that machine learning performs **lossy compression of historical data**:

$$
\text{historical data} \rightarrow \text{learning algorithm} \rightarrow \text{model parameters}
$$

The parameters do not store every training example exactly. They retain patterns useful for the task and discard detail that the objective does not reward. As datasets grow and the relationships to be represented become richer, models often need greater **capacity**—commonly more parameters and layers—to capture those patterns. This is a motivation rather than a fixed rule: parameter needs also depend on the architecture, task, data quality and efficiency of the representation.

| Traditional machine learning | Deep learning |
|---|---|
| Often relies on manually designed features | Learns multiple layers of representations |
| Often effective with smaller, structured datasets | Often excels with large, unstructured datasets |
| Usually cheaper to train and easier to interpret | Can require substantial data and computation |
| Includes linear models, trees and boosting ensembles | Includes CNNs, RNNs and Transformers |

> Deep learning is not automatically the best choice. Prefer a simpler model when it meets the objective with less data, computation and complexity.


### 2.3 Why deep learning became successful

![Algorithms update network weights from errors, big data supplies many input-output examples, and a GPU processes matrix operations in parallel](assets/why-deep-learning.svg){fig-alt="Three panels show error-driven weight updates, many historical input-output pairs, and parallel GPU matrix processing" style="width: 100%; height: auto;"}

- **Algorithms:** backpropagation, improved activation functions, optimisation methods and model architectures made deeper networks easier to train.
- **Big data:** more historical examples exposed useful patterns and supported models with greater capacity.
- **Higher-performance computing:** GPUs, specialised accelerators and distributed training made large-scale calculations practical.

## 3. Representative Deep Learning Architectures

CNNs, RNNs and Transformers are representative **model architectures**, not exclusive application categories. For each: **key features → how it works → suitable tasks**. The same task may use different architectures.

![Deep learning branches into CNNs, RNNs and Transformers, with representative uses shown beneath each architecture](assets/session9-deep-learning-mind-map.svg){fig-alt="Deep learning branches into CNNs, RNNs and Transformers; each branch lists typical uses rather than an exclusive data type" style="width: 100%; height: auto;"}

*Typical uses overlap: Transformers can process images; RNNs have also been used for language models.*

### 3.1 CNNs

**Key features**

- **Local connectivity:** each filter sees a small region.
- **Weight sharing:** the same filter scans every position.
- **Hierarchical features:** edges → textures or parts → objects.

![A CNN maps an input image through local shared filters and deeper feature layers to a task head](assets/session9-cnn-architecture.svg){fig-alt="An input image passes through local shared convolution filters, deeper feature layers and a task head" style="width: 100%; height: auto;"}

**How it works**

- Sliding filters create **feature maps** from local regions.
- Repeated convolution and activation build deeper features.
- Pooling or strides may reduce image size; a task head produces the final output.

**Suitable inputs and tasks:** images and other spatially structured inputs; classification and object detection.

| Suitable scenario | Representative model or component | Why it fits |
|---|---|---|
| Whole-image classification | LeNet-5, AlexNet, VGG, ResNet | Hierarchical spatial features support a whole-image class prediction |
| Object detection | Faster R-CNN, YOLO | Spatial feature maps retain evidence for locating and classifying objects |
| Image generation with convolutional components | DCGAN; convolutional U-Net in a diffusion model | Local filters help model image structure during generation or denoising |

**Image generation:** DCGAN and diffusion U-Nets use convolutional components. A CNN alone is not a generative model.

### 3.2 RNNs, LSTMs and GRUs

**Key feature:** an RNN carries a **hidden state** from one sequence step to the next, so order and earlier context matter.

**How it works**

- A shared cell combines the **current input** with the **previous state**.
- **LSTM:** cell state and gates control what to keep, forget and output.
- **GRU:** update and reset gates control how past and new information combine.

![A shared RNN cell processes inputs in order and passes hidden state from one time step to the next](assets/session9-rnn-architecture.svg){fig-alt="Three recurrent cells with shared weights process ordered inputs and pass hidden state from step to step; LSTM and GRU add gates" style="width: 100%; height: auto;"}

**Limitations:** plain RNNs struggle with long-distance dependencies; LSTMs and GRUs help, but step-by-step computation limits parallelism.

**Suitable inputs and tasks:** ordered sound frames, words and time-series values; speech recognition, sentiment analysis, text generation and Seq2Seq translation.

| Suitable scenario | Representative model or design | Why it fits |
|---|---|---|
| Sentiment or time-series classification | Simple RNN or GRU classifier | Summarises an ordered input into a final prediction |
| Speech recognition | [Deep Speech 2](https://proceedings.mlr.press/v48/amodei16.html), with convolutional and recurrent layers | Processes successive sound frames and their context |
| Translation or text generation | LSTM encoder-decoder; attention-based GRU encoder-decoder | Maps a variable-length input sequence to an output sequence |
| Image captioning | CNN image encoder with an LSTM decoder | Generates an ordered caption from visual features |

### 3.3 Transformers

**Starting point: Seq2Seq.** Input sequence → output sequence (for example, translation). Seq2Seq is a **task structure**, not an architecture: RNNs or Transformers can implement it.

**Problem → key feature**

- Early RNN Seq2Seq: one fixed-length context vector can lose details from long inputs.
- **Attention:** select relevant input positions for each output step.
- **Transformer:** use attention to connect positions without a step-by-step recurrent state.

**How it works**

- Queries compare with keys; attention weights combine values.
- Self-attention connects positions; multiple heads learn different relationships.
- Position information preserves order; feed-forward layers refine representations.

![Token and position representations pass through self-attention and a feed-forward block to a task output](assets/session9-transformer-architecture.svg){fig-alt="A simplified Transformer block combines token and position representations, self-attention and a feed-forward layer before a task output" style="width: 100%; height: auto;"}

The diagram shows one simplified block. Models stack blocks as **encoder-only**, **encoder-decoder** or **decoder-only**. A decoder's causal mask hides later tokens during next-token prediction.

**Suitable inputs and tasks:** text classification, translation, summarisation, language modelling and image patches. Attention connects distant positions and supports parallel input processing.

| Suitable scenario | Representative model and design | Why it fits |
|---|---|---|
| Text classification or contextual representations | BERT, encoder-only | Each token can use context from the full input |
| Translation or summarisation as Seq2Seq | Original Transformer or T5, encoder-decoder | The decoder attends to relevant input positions when producing each output token |
| Autoregressive language modelling and text generation | GPT or Llama family, decoder-only | Causal attention uses preceding tokens to predict the next one |
| Image classification | [Vision Transformer (ViT)](https://arxiv.org/abs/2010.11929), encoder-style | Attention relates image patches across the image |

**Not exclusive:** Seq2Seq can use RNNs or Transformers; image tasks can use CNNs or Vision Transformers.

## 4. Deep Learning Programming Frameworks

A deep learning framework provides tensor operations, automatic differentiation, reusable layers, loss functions, optimisers, data pipelines, accelerator support and model serialisation. These facilities allow practitioners to concentrate on the model and data rather than deriving and implementing every gradient manually.

### 4.1 PyTorch

PyTorch uses `torch.Tensor` for multidimensional data and automatic differentiation to record operations required for gradient calculation. Models commonly inherit from `torch.nn.Module`, while an explicit training loop makes the forward pass, loss calculation, gradient reset, backpropagation and optimiser step visible. This flexibility is useful for learning, research and custom model development.

```python
import torch
from torch import nn

model = nn.Sequential(
    nn.Conv2d(in_channels=1, out_channels=8, kernel_size=3),
    nn.ReLU(),
    nn.Flatten(),
    nn.Linear(in_features=8 * 26 * 26, out_features=10),
)
```

The snippet is illustrative and is not executed in this notebook because PyTorch is an optional dependency.


### 4.2 TensorFlow and Keras

TensorFlow provides tensors, automatic differentiation and tools for training and deployment. Keras is its high-level model API. The `Sequential` API is concise for a simple stack of layers, while the Functional API supports models with branches, multiple inputs or shared layers. Built-in `compile`, `fit` and `evaluate` methods can manage a standard training workflow, and custom loops remain available when more control is required.

```python
import tensorflow as tf

model = tf.keras.Sequential(
    [
        tf.keras.layers.Conv2D(8, kernel_size=3, activation="relu"),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(10),
    ]
)
```

This equivalent model definition is also illustrative and is not executed unless TensorFlow is installed.


### 4.3 Comparing the frameworks

| Area | PyTorch | TensorFlow with Keras |
|---|---|---|
| Tensor type | `torch.Tensor` | `tf.Tensor` |
| Model definition | `nn.Module` or `nn.Sequential` | Sequential, Functional or subclassed model |
| Standard training style | Often an explicit loop | Often `compile` and `fit` |
| Customisation | Direct, flexible Python control | High-level APIs plus custom loops |
| Deployment | PyTorch deployment and serving tools | Broad TensorFlow serving and edge ecosystem |

Neither framework is universally best. Consider the team's existing skills, pretrained-model availability, required deployment target, hardware support, maintainability and the degree of training-loop customisation required.
